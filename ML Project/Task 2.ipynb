```python
#Check missing value
print(df.isnull().sum())
```

```python
#Fill missing value
mod = df["fraud reported"].mode()[0]
df["fraud reported"] = df["fraud reported"].fillna(mod)
print(df.isnull().sum())
```

```python
#CSV file does not contain duplicate rows
print("Duplicate row before removing:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicate row after removing:", df.duplicated().sum())
```

```python
df["claim_date"] = pd.to_datetime(df["claim_date"], format="mixed")
```

```python
#Detect and handle outlier
Q1 = df["annual_income"].quantile(0.25)
Q3 = df["annual_income"].quantile(0.75)

#for extreme value detect
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

#Check how many outliers present
outliers = df[(df["annual_income"] < lower) | (df["annual_income"] > upper)]
print("Number of outliers:", len(outliers))

#Remove outliers
df = df[(df["annual_income"] >= lower) & (df["annual_income"] <= upper)]
print(df.shape)
```

```python
#Histogram
plt.figure(figsize=(8,5))

plt.hist(df["annual_income"], bins=20, color="skyblue", edgecolor="black")

plt.title("Distribution of annual income")
plt.xlabel("Annual Income")
plt.ylabel("Frequency")

plt.show()
```

```python
#Box-plot
plt.figure(figsize=(6,4))

plt.boxplot(df['form defects'].dropna())

plt.title('Box Plot of Form Defects')
plt.ylabel('Form Defects')

plt.show()
```

```python
#Bar chart
df["fraud reported"].value_counts().plot(kind="bar", color=["green","red"])

plt.title("Fraud Reported")
plt.xlabel("Fraud")
plt.ylabel("Count")

plt.show()
```

```python
#Pie chart
df["fraud reported"].value_counts().plot(kind="pie",autopct="%1.1f%%")

plt.title("Fraud Reported")
plt.ylabel("")

plt.show()
```
